# UrbanFlow — Smart City Traffic & Accident Analysis
## 01. Exploratory Geospatial & Temporal Analysis (Mumbai Metropolitan Region)

This notebook explores collision frequencies, weather correlations, spatial distributions, and diurnal traffic patterns across major transit corridors in Mumbai and Navi Mumbai.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="darkgrid")
plt.rcParams['figure.figsize'] = (12, 6)
```

```python
# Load historical accident dataset
df_accidents = pd.read_csv('../datasets/sample/mumbai_accidents_historical.csv')
df_accidents['timestamp'] = pd.to_datetime(df_accidents['timestamp'])
print(f"Total incidents loaded: {len(df_accidents)}")
df_accidents.head()
```

```python
# Severity distribution
print(df_accidents['severity'].value_counts(normalize=True))
df_accidents['severity'].value_counts().plot(kind='bar', color=['#38bdf8', '#f59e0b', '#ef4444'])
plt.title('Accident Severity Distribution across Mumbai Corridors')
plt.ylabel('Recorded Collisions')
plt.show()
```

```python
# Hourly diurnal distribution
df_accidents['hour'] = df_accidents['timestamp'].dt.hour
hourly_counts = df_accidents.groupby('hour')['incident_id'].count()

plt.figure(figsize=(12, 5))
plt.plot(hourly_counts.index, hourly_counts.values, marker='o', color='#38bdf8', linewidth=2.5)
plt.axvspan(8, 11, color='#f59e0b', alpha=0.2, label='Morning Peak (08-11h)')
plt.axvspan(17, 21, color='#ef4444', alpha=0.2, label='Evening Peak (17-21h)')
plt.title('24-Hour Diurnal Accident Pattern in Mumbai')
plt.xlabel('Hour of Day (24h)')
plt.ylabel('Incident Frequency')
plt.legend()
plt.show()
```
